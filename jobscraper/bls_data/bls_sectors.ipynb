```python
df = sectors_df.copy()
# sector_types_df = df.groupby(['I_GROUP']).nunique()
sectors_types_df = df[df['I_GROUP'] == 'sector']
sectors_types_df.head(5)
```

```python
## National Employment by Sector 
occur_df = sectors_types_df.copy()
occur_df['TOT_EMP'] = pd.to_numeric(occur_df['TOT_EMP'], errors='coerce')
occur_df = occur_df.groupby('NAICS_TITLE')['TOT_EMP'].sum().reset_index()
occur_df = occur_df.sort_values(by='TOT_EMP', ascending=False)
occur_df
```

```python
## Massachusetts Employment by Sector 
ma_emp_df = sectors_types_df[sectors_types_df['AREA_TITLE'] == 'Massachusetts'].copy()
ma_emp_df['TOT_EMP'] = pd.to_numeric(ma_emp_df['TOT_EMP'], errors='coerce')
ma_emp_df = ma_emp_df.groupby('NAICS_TITLE')['TOT_EMP'].sum().reset_index()
ma_emp_df = ma_emp_df.sort_values(by='TOT_EMP', ascending=False)
ma_emp_df
```
